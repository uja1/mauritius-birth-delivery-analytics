```python

# Calculate delivery and birth-related rates

df_clean["CSection_Rate"] = (
    df_clean["Caesarean"] / df_clean["Deliveries"] * 100
)

df_clean["Vaginal_Rate"] = (
    df_clean["Vaginal"] / df_clean["Deliveries"] * 100
)

df_clean["LBW_Rate"] = (
    df_clean["LBW Births"] / df_clean["Live Births"] * 100
)

df_clean["Stillbirth_Rate"] = (
    df_clean["Stillbirths"] / df_clean["Total Births"] * 1000
)

df_clean["Live_Birth_Rate"] = (
    df_clean["Live Births"] / df_clean["Total Births"] * 1000
)

# Display the results
df_clean[
    [
        "Year",
        "Sector",
        "CSection_Rate",
        "Vaginal_Rate",
        "LBW_Rate",
        "Stillbirth_Rate",
        "Live_Birth_Rate"
    ]
].head(10).round(2)
```

```python

# Summarise total deliveries by year and sector

delivery_trend = df_clean.pivot_table(
    index="Year",
    columns="Sector",
    values="Deliveries",
    aggfunc="sum"
)

display(delivery_trend.head())

# Plot the trend
delivery_trend.plot(
    kind="line",
    marker="o",
    figsize=(12, 6)
)

plt.title("Annual Deliveries by Healthcare Sector (2008–2023)")
plt.xlabel("Year")
plt.ylabel("Number of Deliveries")
plt.legend(title="Healthcare Sector")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

# Calculate average Caesarean-section rate by sector

csection_summary = (
    df_clean.groupby("Sector")["CSection_Rate"]
    .mean()
    .round(2)
)

display(csection_summary)

# Visualise the comparison
csection_summary.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Average Caesarean-Section Rate by Sector")
plt.xlabel("Healthcare Sector")
plt.ylabel("Average Caesarean-Section Rate (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

# Summarise deliveries by period and sector

period_summary = df_clean.pivot_table(
    index="Period",
    columns="Sector",
    values="Deliveries",
    aggfunc="mean"
)

# Keep periods in chronological order
period_order = ["2008-2019 Pre-COVID", "2020-2021 COVID", "2022-2023 Recovery"]
period_summary = period_summary.reindex(period_order)

display(period_summary.round(2))

# Create a comparison chart
period_summary.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Average Annual Deliveries by Period and Healthcare Sector")
plt.xlabel("Period")
plt.ylabel("Average Annual Deliveries")
plt.xticks(rotation=15, ha="right")
plt.legend(title="Healthcare Sector")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

period_csection = df_clean.pivot_table(
    index="Period",
    columns="Sector",
    values="CSection_Rate",
    aggfunc="mean"
)

period_csection = period_csection.reindex(period_order)

display(period_csection.round(2))

period_csection.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Average Caesarean-Section Rate by Period and Sector")
plt.xlabel("Period")
plt.ylabel("Average Caesarean-Section Rate (%)")
plt.xticks(rotation=15, ha="right")
plt.legend(title="Healthcare Sector")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

# Calculate average LBW rate by year and sector

lbw_trend = df_clean.pivot_table(
    index="Year",
    columns="Sector",
    values="LBW_Rate",
    aggfunc="mean"
)

display(lbw_trend.head())

# Plot the trend
lbw_trend.plot(
    kind="line",
    marker="o",
    figsize=(12, 6)
)

plt.title("Low-Birth-Weight Rate by Year and Healthcare Sector")
plt.xlabel("Year")
plt.ylabel("Low-Birth-Weight Rate (%)")
plt.legend(title="Healthcare Sector")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

# Summarise key indicators by healthcare sector

sector_kpis = df_clean.groupby("Sector").agg(
    Total_Deliveries=("Deliveries", "sum"),
    Total_Caesarean=("Caesarean", "sum"),
    Total_Vaginal=("Vaginal", "sum"),
    Total_Live_Births=("Live Births", "sum"),
    Total_LBW_Births=("LBW Births", "sum"),
    Total_Stillbirths=("Stillbirths", "sum"),
    Total_Births=("Total Births", "sum")
)

# Calculate overall rates from totals
sector_kpis["CSection_Rate_Pct"] = (
    sector_kpis["Total_Caesarean"]
    / sector_kpis["Total_Deliveries"] * 100
)

sector_kpis["LBW_Rate_Pct"] = (
    sector_kpis["Total_LBW_Births"]
    / sector_kpis["Total_Live_Births"] * 100
)

sector_kpis["Stillbirths_Per_1000"] = (
    sector_kpis["Total_Stillbirths"]
    / sector_kpis["Total_Births"] * 1000
)

display(sector_kpis.round(2))
```

```python

query = """
SELECT
    Sector,
    SUM(Deliveries) AS total_deliveries,
    ROUND(
        100.0 * SUM(Caesarean) / SUM(Deliveries), 2
    ) AS csection_rate_pct,
    ROUND(
        100.0 * SUM("LBW Births") / SUM("Live Births"), 2
    ) AS lbw_rate_pct
FROM birth_statistics
GROUP BY Sector
ORDER BY total_deliveries DESC;
"""

result = pd.read_sql_query(query, connection)
display(result)
```

```python

from pathlib import Path

# Define the SQL folder and create it if necessary
sql_folder = Path(r"C:\mauritius-healthcare-project\sql")
sql_folder.mkdir(parents=True, exist_ok=True)

# Write our SQL analysis queries into a file
sql_code = """
-- 1. Check dataset coverage
SELECT
    COUNT(*) AS total_records,
    MIN(Year) AS first_year,
    MAX(Year) AS last_year
FROM birth_statistics;

-- 2. Top five years by total deliveries
SELECT
    Year,
    SUM(Deliveries) AS total_deliveries
FROM birth_statistics
GROUP BY Year
ORDER BY total_deliveries DESC
LIMIT 5;

-- 3. Compare indicators by sector
SELECT
    Sector,
    SUM(Deliveries) AS total_deliveries,
    ROUND(100.0 * SUM(Caesarean) / SUM(Deliveries), 2)
        AS csection_rate_pct,
    ROUND(100.0 * SUM("LBW Births") / SUM("Live Births"), 2)
        AS lbw_rate_pct,
    ROUND(1000.0 * SUM(Stillbirths) / SUM("Total Births"), 2)
        AS stillbirths_per_1000
FROM birth_statistics
GROUP BY Sector
ORDER BY total_deliveries DESC;
"""

sql_file = sql_folder / "birth_analysis.sql"
sql_file.write_text(sql_code, encoding="utf-8")

print("SQL file created successfully!")
print("Location:", sql_file)
```

```python

# Compare delivery volumes across the three periods
period_summary = df_clean.groupby(
    ["Period", "Sector"], observed=True
).agg(
    Total_Deliveries=("Deliveries", "sum"),
    Total_Caesarean=("Caesarean", "sum"),
    Total_Live_Births=("Live Births", "sum"),
    Total_LBW_Births=("LBW Births", "sum"),
    Total_Stillbirths=("Stillbirths", "sum"),
    Total_Births=("Total Births", "sum")
).reset_index()

# Calculate weighted rates for each period and sector
period_summary["Caesarean_Rate_%"] = (
    period_summary["Total_Caesarean"]
    / period_summary["Total_Deliveries"] * 100
)

period_summary["LBW_Rate_%"] = (
    period_summary["Total_LBW_Births"]
    / period_summary["Total_Live_Births"] * 100
)

period_summary["Stillbirths_per_1000"] = (
    period_summary["Total_Stillbirths"]
    / period_summary["Total_Births"] * 1000
)

display(period_summary.round(2))
```

```python

# Average annual deliveries by period and sector
annual_delivery_summary = df_clean.groupby(
    ["Period", "Sector"], observed=True
).agg(
    Total_Deliveries=("Deliveries", "sum"),
    Number_of_Years=("Year", "nunique")
).reset_index()

annual_delivery_summary["Average_Annual_Deliveries"] = (
    annual_delivery_summary["Total_Deliveries"]
    / annual_delivery_summary["Number_of_Years"]
)

display(annual_delivery_summary.round(2))
```

```python

# Compare Pre-COVID with Recovery
pre_covid = annual_delivery_summary[
    annual_delivery_summary["Period"] == "2008-2019 Pre-COVID"
].set_index("Sector")

recovery = annual_delivery_summary[
    annual_delivery_summary["Period"] == "2022-2023 Recovery"
].set_index("Sector")

comparison = pd.DataFrame({
    "Pre_COVID_Avg_Annual_Deliveries":
        pre_covid["Average_Annual_Deliveries"],
    "Recovery_Avg_Annual_Deliveries":
        recovery["Average_Annual_Deliveries"]
})

comparison["Change_in_Deliveries"] = (
    comparison["Recovery_Avg_Annual_Deliveries"]
    - comparison["Pre_COVID_Avg_Annual_Deliveries"]
)

comparison["Percentage_Change_%"] = (
    comparison["Change_in_Deliveries"]
    / comparison["Pre_COVID_Avg_Annual_Deliveries"] * 100
)

display(comparison.round(2))
```

```python

from pathlib import Path

project_root = Path(r"C:\mauritius-healthcare-project")

for folder in [
    "data/raw",
    "data/cleaned",
    "notebooks",
    "sql",
    "powerbi",
    "reports",
    "images"
]:
    folder_path = project_root / folder
    print(f"\n{folder}/")

    if folder_path.exists():
        files = list(folder_path.iterdir())

        if files:
            for file in files:
                print("  -", file.name)
        else:
            print("  (folder is empty)")
    else:
        print("  (folder not found)")

print("\nProject root files:")
for file in project_root.iterdir():
    if file.is_file():
        print("  -", file.name)
```
